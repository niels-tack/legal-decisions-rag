# Download pdf files

This scripts iterates over the relevant cases, including the hyperlink where the pdf file is stored. Then it downloads all files into the designated download folder.

```python
import os
import pandas as pd
import wget
import requests
from urllib.request import urlopen
from bs4 import BeautifulSoup
```

```python
# show all outputs of cell, not merely of last line (i.e. default of Jupyter Notebook)
from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = "all"
```

```python
# change working directory + set download location
data_location = "../data"
download_folder = "../data/CC_judgments_pdf"
overview_filename = 'CC_2025-2025.pkl'
```

```python
# read in data file
overview_df = pd.read_pickle(os.path.join(data_location, overview_filename))
overview_df.head()
```

```python
# Add https to the URLS (to errors regarding scheme-less URLs)
overview_df['url'] = overview_df['url'].apply(lambda x: x if x.startswith('https://') else 'https://' + x)
overview_df.head()
```

```python
# download pdfs based on all relevant urls to download_folder
for url in overview_df["url"]:
   wget.download(url, download_folder)
print("{} judgments downloaded".format(len(overview_df["url"])))
```
